| Operation            | Meaning                    |
| -------------------- | -------------------------- |
| `enqueue()`          | Add item to the queue      |
| `dequeue()`          | Remove item from the queue |
| `peek()` / `front()` | View first item            |
| `isEmpty()`          | Check if empty             |
| `size()`             | Total items                |


```python
# queue using list
queue = []

queue.append(1)
queue.append(2)
queue.append(3)

print(queue)

poppedVal = queue.pop(0)
print(poppedVal)
```

```python
# make your own queue
class Queue:
  def __init__(self):
    self.items = []
  def enque(self, item):
    self.items.append(item)
  def deque(self):
    self.items.pop(0)
  def front(self):
    frontVal = self.items[0]
    print(frontVal)
  def isEmpty(self):
    length = len(self.items)==0
    print(length)
  def view(self):
    for item in self.items:
      print(f'{item}->', end="")
    print()
queue = Queue()

queue.enque(1)
queue.enque(2)
queue.enque(3)
queue.view()

queue.front()

queue.deque()
queue.view()

queue.isEmpty()
```
