ordered/ immutable/ allow duplicates/ faster then lists/ used for fixed data, unpacking, returning multiple values, etc.

```python
t1 = (1,2,3) # tuple with parentheses
t2 = 'a','b','d' # tuple without parentheses
single_element_tuple = (10,) # this comma is necessary
print(type(t1), type(t2))

a,b,c = t1 # tuple unpacking
print(a,b,c)

# partial unpacking
t = (1,3,4,5,6,8,9)
x, *y, z = t # *y will return a list
print(x,y,z)

# tuple slicing
print(t[2:5]) # including the first, excluding the last

# tuple as dictionary key
co_ords = {}
co_ords[(1,2)] = 'A'
co_ords[(2,3)] = 'B'
print(co_ords)
```

```python
import sys

# tuple use less memory
print(sys.getsizeof([1,2,3])) # list
print(sys.getsizeof((1,2,3)))
```

```python
# methods of tuple
t = (1,5,9,3,7,6,1,6,1)
print(t.count(1)) # it counts how many times 1 has appeared in the tuple
print(t.index(6)) # it get the 1st index of 6
```

```python
# returning multiple value with tuple
def calc(a,b):
  return a+b, a-b, a*b, a/b # as we have not used any parentheses, that means returns will be a tuple
sumValue, subValue, mulValue, divValue = calc(4,5)

print(sumValue, subValue, mulValue, divValue)
```

```python
# important for ML
pairs = [(1,'b'), (6,'a'), (2,'d')]
pairs.sort() # this will sort our tuple in comparison of 1st element of tuple
print(pairs)

pairs.sort(key=lambda x:x[1]) # this will sort according to 2nd element of tuple
print(pairs)
```

```python
# tuple don't support comprehension, they use generators
t = tuple(x*x for x in range(5)) # must explicitly include tuple keyword, unless it will make an error
print(t)
```
