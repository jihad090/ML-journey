```python
char = ['a','b','e','f']

print(char[0:2]) #list[start:end] include start, exclude end
print(char[2:-1]) #last element index is -1, thats why it excludes 'f' 

char.append('z')
print('add element: ',char)

char.insert(1,'w')
print('Insert \'w\' at index 1: ',char)

char.extend([1,2,3,4]) # add multiple item in previous list
print(char)

char.pop() # remove last element and return it
print(char.pop(3)) # remove item at index 3 and return that item

char.clear()
print('Cleared List: ',char)
```

```python
# traverse list
list = [1,-2,4,-5,3]

# for item in list:
#   print(item)

for index, item in enumerate(list): # get the item and its index
  print(index, item)

list.sort()  # assending sort the list in place
print(list)  # print the sorted list

list.sort(reverse=True)
print(list)
```

```python
# list comprehensions
squares = [x*x for x in range(10)] # generating square numbers
print(squares)

evens = [x for x in range(25) if x%2==0]
print(evens)

pairs = [(x, y) for x in range(3) for y in range(2)]
print(pairs)
```
