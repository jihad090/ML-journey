| Operation          | Description                    |
| ------------------ | ------------------------------ |
| `push()`           | Add an item                    |
| `pop()`            | Remove item                    |
| `peek()` / `top()` | View top item without removing |
| `isEmpty()`        | Check if stack is empty        |
| `size()`           | Get number of elements         |


```python
# stack using python list

stack = []

stack.append(1) # like push
stack.append(2) 
stack.append(3) 

print(stack)

poppedValue = stack.pop()
print(f'Popped Value: {poppedValue}')

print(stack)

topValue = stack[-1]
print(f'Top: {topValue}')
```

```python
# make your own stack
class Stack:
  def __init__(self):
    self.items = []
  def push(self, item):
    self.items.append(item)
  def pop(self):
    poppedValue = self.items.pop()
    print(poppedValue) 
  def top(self):
    topVal = self.items[-1]
    print(topValue)
  def isEmpty(self):
    length = len(self.items) == 0
    print(length)
  def view(self):
    for item in self.items:
      print(f'{item}->', end="")
    print()

stack = Stack()

stack.push(2)
stack.push(3)
stack.push(5)
stack.view()

stack.pop()
stack.view()

stack.isEmpty()

stack.top()
```
