```python
list = [1,6,7,8,9]

list.append(50) #add new value
print(list)

list.insert(1,-9) # add value at certain index
print(list)

list.remove(-9) # here we must ensure that -9 is present in list, else it will throw an error
print(list)

poppedVal = list.pop(2) # pop by index
print(poppedVal, list)

list[0] = -9
print(list)

for idx,item in enumerate(list): # traverse through list
  print(idx,'->',item)
```

```python
# NumPy arrays are:
# fast
# memory-efficient
# vectorized (operations happen in bulk)
# mathematical (matrix operations)

import numpy as np
arr = np.array([1,2,3,4,5,6])
print(arr)

print(arr.shape)
print(arr.size)
print(arr.dtype)

print(arr/2)

slicedArray = arr[1:3] # array slicing, this line only copy the items, the array remains as it is
print(slicedArray)

reshappedArray = arr.reshape(3,2) # this line convert this array into 3*2 matrix
print(reshappedArray)
```

```python
# statistical method, useful for ML (we must use numpy array)

arr = np.array([1,2,3,4,5])
print(arr.sum())
print(arr.mean())
print(arr.min())
print(arr.max())
print(np.median(arr))
print(np.std(arr)) #standerd deviation
```

```python
import numpy as np

_2DArray = np.zeros((3,3)) # this will create an 3x3 array initialized with 0 
                           # similarly if we write np.ones((m,n)), it will initialized with 1
print('2D array with np :\n',_2DArray)

_1Drange = np.arange(0,10,2) # (start, end, step) including start, excluding end, executing step
print('1D array with np :\n',_1Drange)

linespaceArray = np.linspace(1,5,10) # (start, end, quantity) will generate an 1D array of 'quantity' times number between start(include) and end(include)
# this might be useful for sampling purpose
print('Line space :\n',linespaceArray)


randomFloatList = np.random.rand(3,3) # provide 3x3 list of float number from 0->1
print('Randon Float number List : \n', randomFloatList)

randomIntList = np.random.randint(1,100,5) # (start, end, quantity) start->inclusive, end->exclusive, quantity-> how much number to create
print(randomIntList)
```
