Comprehensions are compact, expressive, and fast ways to create collections.


```python
# If as a filter -> General Pattern for if statement [expression for item in iterable if condition]
evenSquares = [x*x for x in range(10) if x%2==0]
print(evenSquares)

# General pattern for if-else statement [expression-1 if condition else expression-2 for item in iterable]
result = ['even' if x%2==0 else 'odd' for x in range(10)]
print(result)

# Nested Loop
nested = [(x,y) for x in range(3) for y in range(2)]
print(nested)
```
