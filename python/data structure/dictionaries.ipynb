Python’s most powerful built-in data structure for mapping keys → values.

```python
# Key must be int, float, string, tuple of immutable, frozenset(build in immutable set)
# key cannot be mutable things like list, set, dict, tuple containing list
dict = {'name':'Jihad', 'age':'23', 'address':'Ctg'}
print(dict['name'])
dict['name']='Fahad' # mutable
print(dict)
dict['country']='BD' # adding new key:val pair in dict
print(dict)
dict.pop('age') # ramove age from dict / we can use -> del dict['age']
print(dict)

print(dict.get('salary',0)) # if the salary do not exist it will show the default value, this is a safe way that will not generate an error
contactInfo = {'mobile':'018...', 'whatsApp':'015...'}
dict.update(contactInfo)
print(dict)
```

```python
# iteration technique for dictionary  items() / keys() / values()
dict = {'k1':'v1', 'k2':'v2', 'k3':'v3'}

for key, val in dict.items():
  print(key, val)

for val in dict.values():
  print(val)
```

```python
# so important for ML
squares = { x:x*x for x in range(10) if x*x*x < 70}
print(squares)

# nested structure
student = {
  'name':'Rakib',
  'age': 23,
  'marks': {'math':96, 'phy':95, 'chem':95}
}
print(student['marks']['math'])
print(student.items()) # it provide a list of key value pair
```

```python
# Python has no switch-case
# Dictionary as switch-case
operations = {
  'add': lambda a, b: a+b,
  'sub': lambda a, b: a-b,
  'mul': lambda a, b: a*b 
}
print(operations['add'](3,4))
```

```python
# Shallow copy
a = {'x':[1,2,3]}
b = a.copy() # this will create a shallow copy of a, that means any change in b, will reflect in a also
b['x'].append(4)
print(a,b)

# to overcome this behaviour we need to import and use copy
import copy
p = {'z':[1,2,3]}
q = copy.deepcopy(p)
q['z'].append(9) # this change will not reflect in p
print(p,q)
```

```python
from collections import defaultdict
freq = defaultdict(int)
for ch in "banana":
    # print(ch)
    freq[ch] += 1
print(freq)

# best for counting values quickly
from collections import Counter
c = Counter("banana")
print(c)
```
