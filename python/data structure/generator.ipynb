Generator expressions are lazy, memory efficient iterators that produce values on demand, not all at once. It looks like list but behave very different

```python
# it's declaration is like comprehensation but use () instead of []
# produce values one at a time. doesnot store whole result in memory
# generators are not subscriptable. 

gen = (x for x in range(5)) # 0,1,2,3,4
print(gen) # this will not be helpful, because of the behaviour of generator

print(next(gen))  # 0 
print(next(gen))  # 1
print(next(gen))  # 2

print(list(gen))  #[3, 4]
print(list(gen))  #[]  
# this is happenning because generators are ONE TIME USE, if the value is traversed, then it will never be able to see again, because it is permanantely deleted, and thats why it is memory efficient
```

| Feature    | List Comprehension | Generator Expression |
| ---------- | ------------------ | -------------------- |
| Memory     | High               | Very low             |
| Evaluation | Immediate          | Lazy                 |
| Stored     | Yes                | No                   |
| Reusable   | Yes                | ❌ No (one-time)      |
